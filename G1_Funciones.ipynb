# Problema 6

La temperatura de un servidor (en °C) puede ser modelada mediante la función

$$T(t) = -0,5 t^{2} + 3 t + 20$$

donde $t$ corresponde al tiempo transcurrido (en horas) en un día laboral. Considere que la jornada laboral comienza a las 08:00 horas y termina a las 17:00 horas.

1.  Defina variable dependiente e independiente, indicando unidad de medida.
2.  Determine el dominio contextualizado del problema.
3.  Grafique la función utilizando *Python*.
4.  Mediante un análisis gráfico, estime cuándo el servidor alcanza la máxima temperatura y calcule cuánto es.
5.  Determine la temperatura del servidor a las 13:00 horas y al finalizar la jornada laboral.


```python
#VD: Temperatura
#DOMINIO CONTEX: [0 , 9] el 0 representa 8 y 9 las 17:00

import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def T(t):
    return -0.5*t**2+3*t+20
t = np.arange(0, 9, 0.1)
plt.plot(t, T(t), label ="T(t)")
plt.title('Temperatura en funcion del tiempo')
plt.ylabel('Temperatura (Celsius)')
plt.xlabel('Tiempo (horas)')
plt.legend()
plt.grid()
ax=plt.gca()
ax.spines[["left","bottom"]].set_position(("data",0))
ax.spines[["right","top"]].set_visible(False)
plt.ylim(0,25)
plt.show()
print(f"A las 3 horas se observa que la temperatura maxima es {T(3)}")
print(f"A las 13:00 horas se observa que la temperatura maxima es {T(5)}")
print(f"A las 17:00 horas se observa que la temperatura maxima es {T(9)}")
```

# Problema 7

El número de usuarios de una red social se puede modelar mediante la función:

$$U(t) = \frac{1.000}{1+9e^{-0.5t}}$$

donde $t$ corresponde al tiempo transcurrido en meses.

1.  Defina variable dependiente e independiente, indicando unidad de medida.
2.  Determine la cantidad de usuarios transcurridos $12$ meses.
3.  Grafique la función, utilizando la librería *Matplotlib*, para los primeros dos años de funcionamiento.
4.  ¿Cuánto tiempo debe pasar para que la red social llegue a $800$ usuarios?

```python
#VD: N de usuarios en una red social (Unidades)
#VI: Tiempo (meses)



import matplotlib.pyplot as plt
import numpy as np

e=np.e
def U(t):
    return 1000/(1+9*e**(-0.5*t))
    
print(f"La cantidad de usuarios a los 12 meses es {U(12):.0f} usuarios")

t = np.arange(0, 24, 0.1)
plt.plot(t, U(t), label ="U(t)")
plt.title('Usuarios en funcion del tiempo')
plt.ylabel('Usuarios')
plt.xlabel('Tiempo (meses)')
plt.legend()
plt.grid()

plt.show()
```

```python
t=sp.Symbol('t') #'variable de la ecuación
ecuacion = sp.Eq(U(t), 800) #ecuacion
solucion = sp.solve(ecuacion, t)
print(solucion)

print(U(solucion[0]))
print(f"Para 800 usuarios seran {solucion[0]:.1f} meses, aproximadamente")
```

# Problema 8

El tiempo de ejecución de un algoritmo de ordenación (en segundos) se puede modelar mediante la función:

$$A(n) = 0,01n^{2} + 0,5n + 2$$

donde $n$ corresponde a la cantidad de elementos a ordenar. El algoritmo funciona desde las 23:00 horas hasta las 06:00 horas y debe ordenar $1.562$ elementos.

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Determine el dominio contextualizado de la función.
3. Determine el tiempo de ejecución para ordenar $1.200$ elementos.
4. Grafique, utilizando la biblioteca *Matplotlib*, la función $A(n)$.
5. ¿Cuántos elementos ordena luego de $6$ horas de funcionamiento?

```python
#VD: la variable dependiente es el tiempo de ejecucion de el algoritmo (segundos):V
#VI: cantidad de elementos
#dominio contextualizado [0, 1562] asociado a la variable independiente


def A(n):
    return 0.01*n**2+0.5*n+2
print(f"el tiempo de ejecucion para 1200 es {A(1200)} segundos")
n = np.arange(0,1562,0.1)
plt.plot(n, A(n), label ="A(n)")
plt.title('Tiempo de ejecucion en funcion del n° de elementos')
plt.ylabel('Tiempo de ejecucion en seg')
plt.xlabel('N° de elementos')
plt.legend()
plt.grid()

#Las 6 horas son el resultado, pero hay que expresarlas en segundos

tiempo=6*60*60
print(tiempo)

#Debemos resolver la funcion A(n)=21600

n=sp.Symbol('n') #'variable de la ecuación
ecuacion = sp.Eq(A(n), 21600) #ecuacion
solucion = sp.solve(ecuacion, n)
print(solucion)
print(A(solucion[0]))
print(f"A las 6 horas ordena {solucion[1]:.0f} elementos, aproximadamente.")

```

# Problema 9

El consumo de energía (en $kWh$) de un *Data Center* puede ser modelado por la función:

 $$E(t) = 50\cdot \log(t + 1) + 200$$

donde $t$ corresponde al tiempo (en horas) desde el inicio del monitoreo.

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Determine el consumo de energía del *Data Center* después de $5$ horas.
3. ¿Luego de cuántas horas el consumo será de $350$ kWh?
4. Grafique, utilizando la biblioteca *Matplotlib*, la función $E(t)$.

```python
#VI tiempo desde el inicio del monitoreo
#VD Consumo de energia en el data center

import numpy as np
import matplotlib.pyplot as plt
import sympy as sp


def E(t):
    return 50*np.log10(t + 1) + 200
print(f"El consumo de energia del data center en 5 horas sera de {E(5):.1f} kWh")

t_plot = np.arange(0, 350, 0.1)
plt.plot(t_plot, E(t_plot), label="E(t)")
plt.title('Consumo de energia del data center')
plt.ylabel('Energia (kWh)')
plt.xlabel('Tiempo (horas)')
plt.legend()
plt.grid()
plt.show()

t = sp.Symbol('t', positive=True)
ecuacion = sp.Eq(50*sp.log(t + 1) + 200, 350)
solucion = sp.solve(ecuacion, t)
print(solucion)
print(f"El consumo sera de 350 kWh luego de {float(solucion[0]):.2f} horas")
```
